## Introduction to Python Modules and Scripting

In Python, code organization and reuse are _**essential**_ skills for writing maintainable programs.
This notebook introduces you to several key concepts:

# Libraries, Packages, and Modules

* **Module**: A single Python file (.py) containing functions, classes, and variables. 
  Think of a module as a toolbox with specific tools inside.
  Examples: math.py, random.py

* **Package**: A collection of related modules organized in a directory with a special __init__.py file.
  Think of a package as a toolbox with multiple compartments (modules).
  Examples: numpy, pandas

* **Library**: A collection of packages and modules that provide related functionality.
  Think of a library as an entire workshop with many specialized toolboxes.
  Examples: The "Standard Library" (built-in modules like math, os, sys), or third-party libraries like requests

### File Operations

Reading from and writing to files is a common task in Python:

#### _Reading a file_
```python
# This opens a file, reads it, and closes it again.  The content is then available in the `content` variable
with open('input.txt', 'r') as file:
    content = file.read()
```

#### _Writing to a file_
```python
# This opens a file, writes to it, and closes it again. If the file doesn't exist, it is created.  (Note: if it exists, `write` overwrites what's in there)
with open('output.txt', 'w') as file:
    file.write('Hello, world!')
```

#### _Appending to a file_
```python
# This opens a file, appends to it, and closes it again. If the file doesn't exist, it is created.  (Note: if it exists, `write` overwrites what's in there)
with open('output.txt', 'a') as file:
    file.write('Hello, world!')
```

### Putting it all together

````python
print("🐍 Welcome to the Python Module Zoo 🐍")

# A small dictionary of animal sounds
animal_sounds = {
    'snake': 'hiss',
    'monkey': '',
    'lion': '',
    'duck': 'quack'
}

def speak_animal(animal):
    """Make an animal speak using our dictionary"""
    if animal.lower() in animal_sounds:
        return f"The {animal} says: {animal_sounds[animal.lower()]}!"
    else:
        return f"Sorry, the {animal} is sleeping right now."

# Try our animal sound function. What does the monkey say...?
print(speak_animal('snake'))

# Now let's make a more interesting animal function
def create_animal_sentence(animal, action):
    """Create a silly sentence about an animal"""
    if animal.lower() in animal_sounds:
        sound = animal_sounds[animal.lower()]
        return f"The {animal} says '{sound}' while {action}!"
    else:
        return f"The mysterious {animal} silently {action}."

# Try our sentence function
print(create_animal_sentence('lion', 'jumping over the lazy dog'))
print(create_animal_sentence('duck', 'solving differential equations'))
````

### Reading and Writing `csv` files

So far we've worked with `.txt` files.  But most of the time when we work with data we'll be working with `.csy` files rather than `.txt` ones.  The syntax for working with `.csv` is very similar to working with `.txt` files but with a couple of key differences.

Let’s read the [documentation](https://docs.python.org/3/library/csv.html#) for the `csv` library to work out how we would read and write to a `.csv` file.
- How can we read a row from a `.csv` file?
- How can we write a new row to an existing `.csv` file?
- How can we modify the existing rows of a `.csv` file?

#### _Reading a file_
```python
# This opens a file, reads it, and closes it again.  The content is then available in the `content` iterable

import csv

with open('grades.csv', 'r', newline='') as file:
    content = csv.reader(file, delimiter=' ', quotechar='|')
    for row in content:
        print(row)
```

#### _Writing to a file_
```python
# This opens a file, writes to it, and closes it again. If the file doesn't exist, it is created.  (Note: if it exists, `write` overwrites what's in there)

import csv

data = [
    ["Name", "Age", "Grade", "Subject"],
    ["Alice", 14, "A", "Math"],
    ["Bob", 15, "B", "Science"],
    ["Charlie", 14, "C", "History"],
    ["Diana", 15, "A", "English"],
    ["Ethan", 14, "B", "Math"]
]

with open("students.csv", mode="w", newline="") as file:
    writer = csv.writer(file)
    writer.writerows(data)
```

#### _Appending to a file_
```python
# This opens a file, appends to it, and closes it again. If the file doesn't exist, it is created.  (Note: if it exists, `write` overwrites what's in there)

import csv

with open("students.csv", mode="a", newline="") as file:
    writer = csv.writer(file)
    writer.writerow(["Fred",14,"C", "Geography"])
```

````python
# using just `csv.reader`

import csv

with open('./data/grades.csv', 'r') as file:
    content = csv.reader(file)
    for row in content:
        print(row)
````

````python
# using csv.DictReader allows you to access an individual row; certain columns of the csv data; or access just the column headers
with open('./data/grades.csv', newline='') as file:
    content = csv.DictReader(file)
    for row in content:
        print(row['Name'], row['Age'])

# the header is accessible using the .fieldnames attribute
print(content.fieldnames)

# each row is a dictionary, with the keys being the csv header, and the values being the values for any given row:
print(row)
````

````python
# Similarly, `csv.DictWriter` allows you to create a csv using dictionary rather than list format

with open('more_students.csv', 'w', newline='') as csvfile:
    fieldnames = ["Name", "Age", "Grade", "Subject"]
    writer = csv.DictWriter(csvfile, fieldnames=fieldnames)

    writer.writeheader()
    writer.writerow({"Name": "Alice", "Age": 14, "Grade": "A", "Subject": "Math"})
    writer.writerow({"Name": "Bob", "Age": 15, "Grade": "B", "Subject": "Science"})
    writer.writerow({"Name": "Charlie", "Age": 14, "Grade": "C", "Subject": "History"})
    writer.writerow({"Name": "Diana", "Age": 15, "Grade": "A", "Subject": "English"})
    writer.writerow({"Name": "Ethan", "Age": 14, "Grade": "B", "Subject": "Math"})
````

````python
with open('more_students.csv', mode='a', newline='') as file:
    fieldnames = ["Name", "Age", "Grade", "Subject"]
    writer = csv.DictWriter(file, fieldnames=fieldnames)
    writer.writerow({"Name": "Fred", "Age": 14, "Grade": "C", "Subject": "Geography"})
````
