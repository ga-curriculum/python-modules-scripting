## Introduction to Python Modules and Scripting

In Python, code organization and reuse are _**essential**_ skills for writing maintainable programs.
This notebook introduces you to several key concepts:

# Libraries, Packages, and Modules

* **Module**: A single Python file (.py) containing functions, classes, and variables. 
  Think of a module as a toolbox with specific tools inside.
  Examples: math.py, random.py

* **Package**: A collection of related modules organized in a directory with a special __init__.py file.
  Think of a package as a toolbox with multiple compartments (modules).
  Examples: numpy, pandas

* **Library**: A collection of packages and modules that provide related functionality.
  Think of a library as an entire workshop with many specialized toolboxes.
  Examples: The "Standard Library" (built-in modules like math, os, sys), or third-party libraries like requests

### File Operations

Reading from and writing to files is a common task in Python:

#### _Reading a file_
```python
# This opens a file, reads it, and closes it again.  The content is then available in the `content` variable
with open('input.txt', 'r') as file:
    content = file.read()
```

#### _Writing to a file_
```python
# This opens a file, writes to it, and closes it again. If the file doesn't exist, it is created.  (Note: if it exists, `write` overwrites what's in there)
with open('output.txt', 'w') as file:
    file.write('Hello, world!')
```

#### _Appending to a file_
```python
# This opens a file, appends to it, and closes it again. If the file doesn't exist, it is created.  (Note: if it exists, `write` overwrites what's in there)
with open('output.txt', 'a') as file:
    file.write('Hello, world!')
```

````python
# Fun Exercise: The Python Zoo
# -----------------------------
print("🐍 Welcome to the Python Module Zoo! 🐍")

# A small dictionary of animal sounds
animal_sounds = {
    'snake': 'hiss',
    'monkey': 'ooh ooh aah aah',
    'lion': 'roar',
    'duck': 'quack'
}

def speak_animal(animal):
    """Make an animal speak using our dictionary"""
    if animal.lower() in animal_sounds:
        return f"The {animal} says: {animal_sounds[animal.lower()]}!"
    else:
        return f"Sorry, the {animal} is sleeping right now."

# Try our animal sound function
print(speak_animal('snake'))
print(speak_animal('monkey'))

# Now let's make a more interesting animal function
def create_animal_sentence(animal, action):
    """Create a silly sentence about an animal"""
    if animal.lower() in animal_sounds:
        sound = animal_sounds[animal.lower()]
        return f"The {animal} says '{sound}' while {action}!"
    else:
        return f"The mysterious {animal} silently {action}."

# Try our sentence function
print(create_animal_sentence('lion', 'jumping over the lazy dog'))
print(create_animal_sentence('duck', 'solving differential equations'))
````

## Exercise 3: Using Python Scripts to Automate Tasks

In this section, you'll create functions that could be used in scripts to automate common tasks.
1.  Create a function that generates a report of random numbers.
    - The function should take a parameter 'count' which determines how many random numbers to generate.
    - It should return a list of 'count' random integers between 1 and 100.
2. Create a function that calculates statistics from a list of numbers
    - The function should return a dictionary with the following keys: 'mean', 'median', 'minimum', 'maximum', and 'sample_size'

````python
# 3.1: Create a function that generates a report of random numbers
def generate_random_report(count):
    """Generate a list of 'count' random integers between 1 and 100"""
    return [rd.randint(1, 100) for _ in range(count)]

# Test with 10 random numbers
random_numbers = generate_random_report(10)
print("Generated random numbers:")
print(random_numbers)
````

````python
# 3.2: Create a function that calculates statistics from a list of numbers
def calculate_statistics(numbers):
    """Calculate and return statistics about the numbers"""
    if not numbers:
        return {
            'mean': None,
            'median': None,
            'minimum': None,
            'maximum': None,
            'sample_size': 0
        }
    
    # Sort the numbers for calculating median
    sorted_numbers = sorted(numbers)
    n = len(sorted_numbers)
    
    # Calculate median
    if n % 2 == 0:  # Even number of elements
        median = (sorted_numbers[n//2 - 1] + sorted_numbers[n//2]) / 2
    else:  # Odd number of elements
        median = sorted_numbers[n//2]
    
    # Return statistics dictionary
    return {
        'mean': sum(numbers) / len(numbers),
        'median': median,
        'minimum': min(numbers),
        'maximum': max(numbers),
        'sample_size': len(numbers)
    }

# Test with a list of numbers
test_numbers = [4, 7, 2, 9, 3, 5, 1, 8, 6]
stats = calculate_statistics(test_numbers)
print("Statistics:")
for key, value in stats.items():
    print(f"{key.capitalize()}: {value}")
````

## Exercise 4: Reading and Writing Files

In this section, you'll practice working with files in Python.

1. Create a function that writes the given data to a file
    - The function should take two parameters:
        - `data`: a list of strings
        - `filename`: the name of the file to write to
    - Each string in data should be written to a new line in the file.

2. Create a function that reads the contents of a file
    - The function should take one parameter:`filename`
    - It should return a list of strings, where each string is a line from the file.
    - Make sure to handle potential errors (e.g., file not found).

3. Create a function that appends a single line to an existing file
    - The function should take two parameters:
        - `line`: a string to append
        - `filename`: the name of the file to append to

````python
# 4.1: Create a function that writes the given data to a file
def write_lines_to_file(data, filename):
    """Write each string in data to a new line in the file"""
    with open(filename, 'w') as file:
        for line in data:
            file.write(f"{line}\n")
    print(f"Successfully wrote {len(data)} lines to {filename}")

# Test the function
test_data = ["Line 1", "Line 2", "Line 3", "Line 4"]
write_lines_to_file(test_data, "./sample-solutions/test_output.txt")
````

````python
# 4.2: Create a function that reads the contents of a file
def read_lines_from_file(filename):
    """Read and return all lines from the file"""
    try:
        with open(filename, 'r') as file:
            # Strip newline characters from each line
            return [line.strip() for line in file.readlines()]
    except FileNotFoundError:
        print(f"Error: File '{filename}' not found.")
        return []
    except Exception as e:
        print(f"Error reading file: {e}")
        return []

# Test the function
lines = read_lines_from_file("./sample-solutions/test_output.txt")
print("Lines read from file:")
for i, line in enumerate(lines, 1):
    print(f"{i}. {line}")
````

````python
# 4.3: Create a function that appends a single line to an existing file
def append_line_to_file(line, filename):
    """Append the line to the file"""
    with open(filename, 'a') as file:
        file.write(f"{line}\n")
    print(f"Successfully appended to {filename}")

# Test the function
append_line_to_file("This is an appended line", "./sample-solutions/test_output.txt")
lines = read_lines_from_file("./sample-solutions/test_output.txt")
print("\nLines after appending:")
for i, line in enumerate(lines, 1):
    print(f"{i}. {line}")
````

## Challenge: Combine Everything

Create a script that:
1. Generates 100 random numbers
2. Calculates statistics for these numbers
3. Writes both the numbers and the statistics to files
4. Reads the files back and verifies the statistics

````python
def main():
    """Combines all functionality from the exercises"""
    print("Generating random numbers...")
    numbers = generate_random_report(100)
    
    print("Calculating statistics...")
    stats = calculate_statistics(numbers)
    
    print("Writing data to files...")
    # Convert numbers to strings for file writing
    number_strings = [str(num) for num in numbers]
    write_lines_to_file(number_strings, "random_numbers.txt")
    
    # Convert stats dictionary to strings for file writing
    stats_strings = [f"{key}: {value}" for key, value in stats.items()]
    write_lines_to_file(stats_strings, "statistics.txt")
    
    print("Reading files back...")
    read_numbers = read_lines_from_file("random_numbers.txt")
    read_stats = read_lines_from_file("statistics.txt")
    
    print("Verifying data...")
    # Convert read numbers back to integers
    converted_numbers = [int(num) for num in read_numbers]
    
    # Calculate statistics again to verify
    verification_stats = calculate_statistics(converted_numbers)
    
    print("Statistics verification:")
    for key, value in verification_stats.items():
        print(f"{key}: {value}")
    
    print("Adding timestamp to statistics file...")
    timestamp = f"Generated on: {datetime.now().strftime('%Y-%m-%d %H:%M:%S')}"
    append_line_to_file(timestamp, "statistics.txt")
    
    print("Process completed successfully!")
    return {
        "numbers": numbers[:5] + ["..."] + numbers[-5:],  # Show first and last 5 numbers
        "statistics": stats
    }

# Run the main function
result = main()

# Display a sample of the results
print("\nSample of generated numbers:")
print(result["numbers"])

print("\nFinal statistics:")
for key, value in result["statistics"].items():
    print(f"{key.capitalize()}: {value}")
````

## Bonus: Creating and Using Your Own Module

In this bonus exercise, let's create a simple Python module that we can import and use.

1. Create a new file named `myutils.py` in the same directory as this notebook
2. Add some utility functions to the file
3. Import and use these functions in this notebook


First, let's create a simple module called `myutils.py` with some utility functions:

````python
%%writefile myutils.py
"""
My Utilities Module

This module contains utility functions for common tasks.
"""

def celsius_to_fahrenheit(celsius):
    """
    Convert Celsius temperature to Fahrenheit.
    
    Args:
        celsius (float): Temperature in Celsius
    
    Returns:
        float: Temperature in Fahrenheit
    """
    return (celsius * 9/5) + 32

def fahrenheit_to_celsius(fahrenheit):
    """
    Convert Fahrenheit temperature to Celsius.
    
    Args:
        fahrenheit (float): Temperature in Fahrenheit
    
    Returns:
        float: Temperature in Celsius
    """
    return (fahrenheit - 32) * 5/9

def format_name(first_name, last_name):
    """
    Format a name in 'Last, First' format.
    
    Args:
        first_name (str): First name
        last_name (str): Last name
    
    Returns:
        str: Formatted name
    """
    return f"{last_name.title()}, {first_name.title()}"

# Constants
PI = 3.14159265359
GOLDEN_RATIO = 1.61803398875

# Version information
__version__ = "0.1.0"
````

````python
# Now let's import and use our module
import myutils

# Display module help information
help(myutils)
````

````python
# Test the functions in our module
print(f"Module version: {myutils.__version__}\n")

# Test temperature conversion
celsius_temp = 25
fahrenheit_temp = myutils.celsius_to_fahrenheit(celsius_temp)
print(f"{celsius_temp}°C = {fahrenheit_temp}°F")

converted_back = myutils.fahrenheit_to_celsius(fahrenheit_temp)
print(f"{fahrenheit_temp}°F = {converted_back}°C\n")

# Test name formatting
print(f"Formatted name: {myutils.format_name('john', 'doe')}\n")

# Show constants
print(f"PI constant: {myutils.PI}")
print(f"Golden Ratio: {myutils.GOLDEN_RATIO}")
````
